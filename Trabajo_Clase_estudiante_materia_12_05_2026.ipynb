```
class Estudiante:
    """
    Clase que representa a un estudiante con sus datos personales y académicos
    """

    def __init__(self, nombre, apellido, id_estudiante, carrera, semestre, telefono, direccion):
        """
        Constructor de la clase Estudiante
        """
        self.nombre = nombre
        self.apellido = apellido
        self.id = id_estudiante
        self.carrera = carrera
        self.semestre = semestre
        self.telefono = telefono
        self.direccion = direccion

    def corregir_nombre(self, nuevo_nombre):
        """
        Método para corregir el nombre del estudiante
        """
        self.nombre = nuevo_nombre
        print(f"✓ Nombre actualizado a: {self.nombre}")

    def corregir_apellido(self, nuevo_apellido):
        """
        Método para corregir el apellido del estudiante
        """
        self.apellido = nuevo_apellido
        print(f"✓ Apellido actualizado a: {self.apellido}")

    def corregir_carrera(self, nueva_carrera):
        """
        Método para corregir la carrera del estudiante
        """
        self.carrera = nueva_carrera
        print(f"✓ Carrera actualizada a: {self.carrera}")

    def imprimir_info_estudiante(self):
        """
        Método para imprimir toda la información del estudiante
        """
        print("\n" + "="*50)
        print("INFORMACIÓN DEL ESTUDIANTE")
        print("="*50)
        print(f"Nombre: {self.nombre}")
        print(f"Apellido: {self.apellido}")
        print(f"ID: {self.id}")
        print(f"Carrera: {self.carrera}")
        print(f"Semestre: {self.semestre}")
        print(f"Teléfono: {self.telefono}")
        print(f"Dirección: {self.direccion}")
        print("="*50 + "\n")


# ====== PROGRAMA INTERACTIVO ======
print("🎓 SISTEMA DE REGISTRO DE ESTUDIANTES\n")

# Solicitar datos al usuario
print("Por favor, ingrese los datos del estudiante:\n")

nombre = input("Nombre: ")
apellido = input("Apellido: ")
id_estudiante = input("ID: ")
carrera = input("Carrera: ")
semestre = int(input("Semestre: "))
telefono = input("Teléfono: ")
direccion = input("Dirección: ")

# Crear el estudiante con los datos ingresados
estudiante = Estudiante(nombre, apellido, id_estudiante, carrera, semestre, telefono, direccion)

# Mostrar la información
print("\n✓ Estudiante creado exitosamente!")
estudiante.imprimir_info_estudiante()

# Preguntar si desea corregir algo
print("¿Desea corregir algún dato? (si/no)")
corregir = input().lower()

if corregir == "si" or corregir == "sí":
    print("\n¿Qué desea corregir? (Puede seleccionar varias opciones separadas por comas)")
    print("1. Nombre")
    print("2. Apellido")
    print("3. Carrera")
    print("\nEjemplo: 1,3 (para corregir nombre y carrera)")

    opciones = input("\nOpciones: ")

    # Separar las opciones por comas y eliminar espacios
    opciones_lista = [opcion.strip() for opcion in opciones.split(",")]

    print("\n--- Ingrese los nuevos valores ---")

    # Procesar cada opción seleccionada
    for opcion in opciones_lista:
        if opcion == "1":
            nuevo_nombre = input("Nuevo nombre: ")
            estudiante.corregir_nombre(nuevo_nombre)
        elif opcion == "2":
            nuevo_apellido = input("Nuevo apellido: ")
            estudiante.corregir_apellido(nuevo_apellido)
        elif opcion == "3":
            nueva_carrera = input("Nueva carrera: ")
            estudiante.corregir_carrera(nueva_carrera)
        else:
            print(f"❌ Opción '{opcion}' no válida, se omite")

    # Mostrar información actualizada
    print("\n📋 Información actualizada:")
    estudiante.imprimir_info_estudiante()
else:
    print("\n✓ No se realizaron cambios")
```

```
class Materia:
    """
    Clase que representa una materia con su profesor y lista de estudiantes
    """

    def __init__(self, nombre_profesor, nombre_materia, max_estudiantes):
        """
        Constructor de la clase Materia
        """
        self.nombre_profesor = nombre_profesor
        self.nombre_materia = nombre_materia
        self.max_estudiantes = max_estudiantes
        self.lista_estudiantes = []
        self.notas = {}

    def matricular_estudiantes(self, estudiante):
        """
        Método para matricular un estudiante en la materia
        """
        # Verificar si hay cupo disponible
        if len(self.lista_estudiantes) >= self.max_estudiantes:
            print(f" No se puede matricular. Cupo máximo alcanzado ({self.max_estudiantes} estudiantes)")
            return False

        # Verificar si el estudiante ya está matriculado
        for est in self.lista_estudiantes:
            if est.id == estudiante.id:
                print(f" El estudiante {estudiante.nombre} {estudiante.apellido} ya está matriculado")
                return False

        # Matricular al estudiante
        self.lista_estudiantes.append(estudiante)
        print(f" Estudiante {estudiante.nombre} {estudiante.apellido} matriculado en {self.nombre_materia}")
        return True

    def nota(self, id_estudiante, calificacion=None):
        """
        Método para asignar o consultar la nota de un estudiante
        """
        # Verificar si el estudiante está matriculado
        estudiante_encontrado = False
        for est in self.lista_estudiantes:
            if est.id == id_estudiante:
                estudiante_encontrado = True
                break

        if not estudiante_encontrado:
            print(f" El estudiante con ID {id_estudiante} no está matriculado")
            return None

        # Si se proporciona una calificación, asignarla
        if calificacion is not None:
            if 0 <= calificacion <= 5.0:
                self.notas[id_estudiante] = calificacion
                print(f" Nota {calificacion} asignada al estudiante con ID {id_estudiante}")
            else:
                print(f" La nota debe estar entre 0 y 5.0")
            return calificacion

        # Si no se proporciona calificación, consultar
        else:
            if id_estudiante in self.notas:
                nota = self.notas[id_estudiante]
                print(f" Nota del estudiante {id_estudiante}: {nota}")
                return nota
            else:
                print(f" El estudiante {id_estudiante} aún no tiene nota asignada")
                return None

    def mostrar_informacion_materia(self):
        """
        Método para mostrar toda la información de la materia
        """
        print("\n" + "="*60)
        print(f"MATERIA: {self.nombre_materia}")
        print("="*60)
        print(f"Profesor: {self.nombre_profesor}")
        print(f"Cupo máximo: {self.max_estudiantes}")
        print(f"Matriculados: {len(self.lista_estudiantes)}/{self.max_estudiantes}")
        print(f"Cupos disponibles: {self.max_estudiantes - len(self.lista_estudiantes)}")

        if self.lista_estudiantes:
            print("\n LISTA DE ESTUDIANTES:")
            print("-" * 60)
            for i, est in enumerate(self.lista_estudiantes, 1):
                nota = self.notas.get(est.id, "Sin nota")
                print(f"{i}. {est.nombre} {est.apellido} | ID: {est.id} | {est.carrera} | Nota: {nota}")
        else:
            print("\n No hay estudiantes matriculados")

        print("="*60 + "\n")


# ====== PROGRAMA INTERACTIVO ======
print(" SISTEMA DE GESTIÓN DE MATERIAS\n")

# Crear la materia
print("=== CREAR MATERIA ===")
nombre_profesor = input("Nombre del profesor: ")
nombre_materia = input("Nombre de la materia: ")
max_estudiantes = int(input("Número máximo de estudiantes: "))

materia = Materia(nombre_profesor, nombre_materia, max_estudiantes)
print(f"\n✓ Materia '{nombre_materia}' creada exitosamente!\n")

# Menú principal
while True:
    print("\n--- MENÚ PRINCIPAL ---")
    print("1. Matricular estudiante")
    print("2. Asignar nota")
    print("3. Consultar nota")
    print("4. Ver información de la materia")
    print("5. Salir")

    opcion = input("\nSeleccione una opción: ")

    if opcion == "1":
        print("\n=== MATRICULAR ESTUDIANTE ===")
        nombre = input("Nombre: ")
        apellido = input("Apellido: ")
        id_est = input("ID: ")
        carrera = input("Carrera: ")
        semestre = int(input("Semestre: "))
        telefono = input("Teléfono: ")
        direccion = input("Dirección: ")

        nuevo_estudiante = Estudiante(nombre, apellido, id_est, carrera, semestre, telefono, direccion)
        materia.matricular_estudiantes(nuevo_estudiante)

    elif opcion == "2":
        print("\n=== ASIGNAR NOTA ===")
        id_est = input("ID del estudiante: ")
        nota = float(input("Nota (0-5.0): "))
        materia.nota(id_est, nota)

    elif opcion == "3":
        print("\n=== CONSULTAR NOTA ===")
        id_est = input("ID del estudiante: ")
        materia.nota(id_est)

    elif opcion == "4":
        materia.mostrar_informacion_materia()

    elif opcion == "5":
        print("\n✓ ¡Hasta luego!")
        break

    else:
        print("❌ Opción inválida")
```

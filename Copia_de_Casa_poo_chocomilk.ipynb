```
class FloristeriaDeCarton:
    def __init__(self):
        # Atributos de ubicación y estructura
        self.ubicacion = ""              # Nuevo
        self.tamaño = ""
        self.material_base = ""
        self.color_fachada = []
        self.puerta = {}
        self.techo = {}
        self.paredes = {"izquierda": "", "derecha": "", "atras": "", "detalles": ""}

        # Atributos del negocio
        self.tipos_flores = []
        self.tipos_plantas = []
        self.num_floristas = 0
        self.num_clientes_dia = 0
        self.num_estantes = 0
        self.num_puntos_hidratacion = 0
        self.jornada_atencion = ""
        self.servicios_especiales = ""

    def capturar_datos(self):
        print("\n--- 🌸 CONFIGURA TU FLORISTERÍA 'EL FLORISTA' 🌸 ---")
        self.ubicacion = input("Ubicación/Dirección de la floristería: ")
        self.tamaño = input("Tamaño del local (Pequeño, Mediano, Grande): ")
        self.material_base = input("Material de construcción: ")

        print("\n--- DISEÑO EXTERIOR ---")
        colores = input("Colores de la fachada (separados por coma): ")
        self.color_fachada = [c.strip() for c in colores.split(",")]

        p_color = input("Color del marco de la puerta: ")
        p_material = input("Material de la puerta: ")
        self.puerta = {"color": p_color, "material": p_material}

        print("\n--- DETALLES DEL TECHO Y PAREDES ---")
        self.techo['material'] = input("Material del techo: ")
        self.techo['decoracion'] = input("¿Tiene plantas colgantes o algún detalle especial?: ")

        self.paredes['izquierda'] = input("Descripción pared izquierda: ")
        self.paredes['derecha'] = input("Descripción pared derecha: ")
        self.paredes['atras'] = input("Descripción pared de fondo (vitrina): ")
        self.paredes['detalles'] = input("Detalles adicionales (letreros, luces): ")

        print("\n--- INVENTARIO Y OPERACIÓN ---")
        flores = input("Flores que vende (ej: Rosas, Lirios, Girasoles): ")
        self.tipos_flores = [f.strip() for f in flores.split(",")]

        plantas = input("Plantas que vende (ej: Helechos, Suculentas, Palmeras): ")
        self.tipos_plantas = [p.strip() for p in plantas.split(",")]

        self.num_floristas = int(input("Número de expertos floristas: "))
        self.num_clientes_dia = int(input("Promedio de clientes por día: "))
        self.num_estantes = int(input("Número de estantes de exhibición: "))
        self.num_puntos_hidratacion = int(input("Número de baños/puntos de agua: "))
        self.jornada_atencion = input("Jornada de atención (ej: 6am - 4pm): ")
        self.servicios_especiales = input("Servicios especiales (ej: Eventos, Domicilios): ")

    def mostrar_descripcion(self):
        print("\n" + "="*65)
        print(f"🌿 NEGOCIO: 'EL FLORISTA' | {self.tamaño.upper()} 🌿")
        print(f"📍 UBICACIÓN: {self.ubicacion}")
        print("="*65)
        print(f" 🏗️  Estructura: {self.material_base} | Fachada: {', '.join(self.color_fachada)}")
        print(f" 🚪 Entrada: Puerta de {self.puerta['material']} color {self.puerta['color']}")
        print(f" 🏠 Techo: {self.techo['material']} con {self.techo['decoracion']}")
        print(f" ✨ Detalles: {self.paredes['detalles']}")
        print("-" * 45)
        print(f" 📋 Catálogo y Personal:")
        print(f"    - 🌷 Flores: {', '.join(self.tipos_flores)}")
        print(f"    - 🍃 Plantas: {', '.join(self.tipos_plantas)}")
        print(f"    - 👥 Staff: {self.num_floristas} floristas")
        print("-" * 45)
        print(f" ⚙️  Operación:")
        print(f"    - Horario: {self.jornada_atencion}")
        print(f"    - Instalaciones: {self.num_estantes} estantes | {self.num_puntos_hidratacion} puntos de agua")
        print(f"    - Servicios: {self.servicios_especiales}")
        print("="*65)

# --- Ejecución ---
mi_floristeria = FloristeriaDeCarton()
mi_floristeria.capturar_datos()
mi_floristeria.mostrar_descripcion()
```

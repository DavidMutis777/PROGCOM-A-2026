```
# -*- coding: utf-8 -*-

class Persona:
    def __init__(self, nombre, cabello, piel, complexion, estatura, prenda_superior, prenda_inferior, calzado):
        self.nombre = nombre
        self.cabello = cabello
        self.piel = piel
        self.complexion = complexion
        self.estatura = estatura
        self.prenda_superior = prenda_superior
        self.prenda_inferior = prenda_inferior
        self.calzado = calzado

    def mostrar_caracteristicas(self):
        print(f"\n===== Características de {self.nombre} =====")
        print(f"  Cabello:          {self.cabello}")
        print(f"  Tono de piel:     {self.piel}")
        print(f"  Complexión:       {self.complexion}")
        print(f"  Estatura:         {self.estatura}")
        print(f"  Prenda superior:  {self.prenda_superior}")
        print(f"  Prenda inferior:  {self.prenda_inferior}")
        print(f"  Calzado:          {self.calzado}")

    def posicion_inicial(self):
        print("Quedarse parado en posición inicial")

    def intercalar_piernas(self, n):
        for i in range(n):
            print("Intercalar pierna izquierda girando hacia un lado")
            print("Intercalar pierna derecha girando hacia un lado")

    def patada(self, n):
        for i in range(n):
            print("Lanzar patada hacia un lado")

    def caminata_con_brincos(self, n):
        for i in range(n):
            print("Caminata con brincos pequeños al ritmo de la música")

    def multiples_patadas(self, n):
        for i in range(n):
            print("Lanzar múltiples patadas hacia un lado seguidas")

    def mover_cabeza_pierna(self, n):
        for i in range(n):
            print("Mover cabeza hacia un lado mientras la pierna se mueve por detrás al ritmo de la música")

    def saltos_piernas(self, n):
        for i in range(n):
            print("Abrir y cerrar piernas dando pequeños saltos hacia adelante con manos agarradas a la cintura")

    def baile_borde_interno(self, n):
        for i in range(n):
            print("Bailar con el borde interno de los pies al ritmo de la música")

    def zombie(self, n):
        for i in range(n):
            print("Agacharse como zombie")
            print("Levantarse como zombie")

    def remate(self):
        print("Pose final de Kung Fu 🥋")

    def kung_fu_fighting(self, n):
        print("=" * 40)
        print(f"  🥋 {self.nombre.upper()} - KUNG FU FIGHTING 🥋")
        print("=" * 40)
        for i in range(n):
            print(f"\n--- Ronda {i+1} ---")
            self.posicion_inicial()
            self.intercalar_piernas(1)
            self.patada(1)
            self.caminata_con_brincos(1)
            self.multiples_patadas(1)
            self.mover_cabeza_pierna(1)
            self.caminata_con_brincos(1)
            self.saltos_piernas(3)
            self.patada(1)
            self.caminata_con_brincos(1)
            self.baile_borde_interno(1)
            self.zombie(2)


# ========== INSTANCIAS ==========

david = Persona("David",
                "Oscuro, corto y ligeramente despeinado",
                "Claro/trigueño",
                "Delgada-media",
                "Media",
                "Polo/camiseta roja",
                "Pantalón blanco",
                "Tenis blancos")

sebastian = Persona("Sebastián",
                    "Oscuro, abundante y un poco largo",
                    "Claro/trigueño",
                    "Robusta",
                    "Media-alta",
                    "Buzo/hoodie negro oscuro",
                    "Jean azul claro",
                    "Tenis claros")

juan_pablo = Persona("Juan Pablo",
                     "Oscuro, medio largo y despeinado",
                     "Claro/trigueño",
                     "Delgada",
                     "Media",
                     "Camiseta/polo verde oscura",
                     "Jean azul oscuro",
                     "Tenis negros con suela blanca")


# ========== CARACTERÍSTICAS ==========

print("=" * 40)
print("  👖 ESTILOS DE VESTIR 👕")
print("=" * 40)

david.mostrar_caracteristicas()
sebastian.mostrar_caracteristicas()
juan_pablo.mostrar_caracteristicas()


# ========== BAILE ==========

david.kung_fu_fighting(1)
david.remate()

sebastian.kung_fu_fighting(1)
sebastian.remate()

juan_pablo.kung_fu_fighting(1)
juan_pablo.remate()
```

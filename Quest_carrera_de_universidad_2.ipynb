```python
import tkinter as tk
from tkinter import messagebox, filedialog

class CalculadoraFinanciera:
    def __init__(self, P, i, n):
        self.P = P  # Capital Presente
        self.i = i / 100  # Tasa de interés
        self.n = n  # Tiempo

    def calcular_interes_simple(self):
        I = self.P * self.i * self.n  # Interés
        F = self.P + I  # Valor Futuro
        return I, F

    def calcular_interes_compuesto(self):
        F = self.P * (1 + self.i) ** self.n  # Valor Futuro
        I = F - self.P  # Interés
        return I, F

class InterfazApp:
    def __init__(self, root):
        self.root = root
        self.root.title("GQ3: Calculadora de Matemáticas Financieras")
        self.root.geometry("400x500")

        # --- Etiquetas y Entradas ---
        tk.Label(root, text="Calculadora de Interés", font=("Arial", 14, "bold")).pack(pady=10)

        tk.Label(root, text="Capital Presente (P):").pack()
        self.entry_p = tk.Entry(root)
        self.entry_p.pack()

        tk.Label(root, text="Tasa de interés anual % (i):").pack()
        self.entry_i = tk.Entry(root)
        self.entry_i.pack()

        tk.Label(root, text="Tiempo en años (n):").pack()
        self.entry_n = tk.Entry(root)
        self.entry_n.pack()

        # --- Botones ---
        tk.Button(root, text="Calcular", command=self.ejecutar_calculo, bg="#4CAF50", fg="white").pack(pady=15)
        tk.Button(root, text="Guardar Resultados", command=self.guardar_en_archivo).pack()

        # --- Área de Resultados ---
        self.text_resultados = tk.Text(root, height=10, width=45)
        self.text_resultados.pack(pady=10)

    def ejecutar_calculo(self):
        try:
            p = float(self.entry_p.get())
            i = float(self.entry_i.get())
            n = float(self.entry_n.get())

            calc = CalculadoraFinanciera(p, i, n)
            i_simple, f_simple = calc.calcular_interes_simple()
            i_comp, f_comp = calc.calcular_interes_compuesto()

            res = (f"--- RESULTADOS ---\n"
                   f"Interés Simple:\n I: ${i_simple:,.2f}\n F: ${f_simple:,.2f}\n\n"
                   f"Interés Compuesto:\n I: ${i_comp:,.2f}\n F: ${f_comp:,.2f}")

            self.text_resultados.delete(1.0, tk.END)
            self.text_resultados.insert(tk.END, res)
        except ValueError:
            messagebox.showerror("Error", "Por favor ingresa valores numéricos válidos.")

    def guardar_en_archivo(self):
        contenido = self.text_resultados.get(1.0, tk.END).strip()
        if not contenido:
            messagebox.showwarning("Advertencia", "No hay resultados para guardar.")
            return

        filepath = filedialog.asksaveasfilename(defaultextension=".txt",
                                                filetypes=[("Archivo de texto", "*.txt")])
        if filepath:
            with open(filepath, "w") as archivo:
                archivo.write(f"Reporte de Cálculo Financiero\n")
                archivo.write(f"Variables usadas: P={self.entry_p.get()}, i={self.entry_i.get()}%, n={self.entry_n.get()}\n")
                archivo.write("-" * 30 + "\n")
                archivo.write(contenido)
            messagebox.showinfo("Éxito", "Resultados guardados correctamente.")

if __name__ == "__main__":
    root = tk.Tk()
    app = InterfazApp(root)
    root.mainloop()
```
